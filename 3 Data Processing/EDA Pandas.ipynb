```sql
%python
conditions = [
loan["person_age"].between(18, 21),
loan["person_age"].between(22, 35),
loan["person_age"].between(36, 50),
loan["person_age"].between(51, 60),
loan["person_age"] > 60
]
choices = [
"Youth",
"Young Adult",
"Adult",
"Elder",
"Pensioner"
]

loan["Age bracket"] = np.select(
conditions,
choices,
default="Unknown"
)
display(loan[["person_age", "Age bracket"]])
```

```sql
%python
loan = loan.rename(columns={"previous_loan_defaults_on_file": "Default Rate"})
display(loan[["Default Rate"]])
```
